```python
from dotenv import load_dotenv
from groq import Groq
import os
load_dotenv()

client = Groq(
    api_key=os.environ.get("GROQ_API_KEY"),  # This is the default and can be omitted
)
```

```python
mensagens = [
    {
        "role": "user",
        "content": "Explique a diferença entre os shopping da Multiplan para a Allos. Responda em português"
    }
]

response = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=mensagens
)

response
```

```python
print(response.choices[0].message.content)
```

# Stream

```python
mensagens = [
    {
        "role": "user",
        "content": "Explique a diferença entre os shopping da Multiplan para a Allos. Responda em português"
    }
]

stream = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=mensagens,
    stream=True
)

for chunk in stream:
    print(chunk.choices[0].delta.content, end="", flush=True)
```

# Transcrição de áudio

```python
import textwrap


def format_text(response):
    text = response.text
    text_formatted = textwrap.fill(text, width=100)
    print(text_formatted)
```

```python
arquivo = 'hello.mp3'

with open(arquivo, 'rb') as audio:
    transcript = client.audio.transcriptions.create(
        file=(arquivo, audio.read()),
        model="whisper-large-v3",
        response_format="json",
        language="pt",
        prompt="Este é um áudio de boas vindas"  # dica para o LLM
    )
    
    format_text(transcript)
```

```python
import textwrap


def format_text(response):
    text = response.content
    text_formatted = textwrap.fill(text, width=100)
    print(text_formatted)
```

```python
completion = client.chat.completions.create(
    model="meta-llama/llama-4-scout-17b-16e-instruct",
    messages=[{
        "role": "user",
        "content": [
            { "type": "text", "text": "O que há nessa imagem?" },
            { "type": "image_url", "image_url": { "url": "https://thumbs.dreamstime.com/b/pessoas-sentadas-no-bar-grupo-de-balc%C3%A3o-bebendo-cerveja-e-conversando-um-com-o-outro-211402266.jpg" } }
        ]
    }],
    temperature=1,
    max_completion_tokens=1024,
    top_p=1,
    stream=False,
    stop=None
)
```

```python
format_text(completion.choices[0].message)
```
