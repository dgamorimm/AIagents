```python
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv())
```

```python
import os
api_key = os.environ.get("GEMINI_API_KEY")
```

```python
import google.generativeai as genai
genai.configure(api_key=api_key)
```

```python
# 2) veja o que sua conta REALMENTE tem disponível
models = [
    m.name for m in genai.list_models()
    if "generateContent" in getattr(m, "supported_generation_methods", [])
]
print("Modelos disponíveis:", models)
```

```python
model = genai.GenerativeModel("gemini-2.5-flash")
```

```python
response = model.generate_content("Qual dica você recomenda para ter uma vida boa?")
```

```python
print(response.text)
```

```python
import textwrap

def format_text(text):
    print(textwrap.fill(text, width=100))
```

```python
format_text(response.text)
```

# Stream

```python
stream = model.generate_content("Qual o sentido da felicidade ?", stream=True)

complete_response = ""
for chunk in stream:
    print(chunk.text, end="", flush=True)
    complete_response += chunk.text
```

# Imagens

```python
import httpx
import os
import base64

model = genai.GenerativeModel("gemini-2.5-pro")
```

```python
image_path = 'https://www.universal-assistance.com/br-blog/wp-content/uploads/2022/09/pontos-turisticos-de-londres-blog-ua-br.jpg'
```

```python
image = httpx.get(image_path)
```

```python
prompt = 'Descreva a imagem em detalhes'
```

```python
response =  model.generate_content([
    {
        'mime_type' : 'image/jpeg',
        'data' : base64.b64encode(image.content).decode('utf-8')
    },
    prompt
])
print(response.text)
```
