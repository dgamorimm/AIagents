```python
from transformers import AutoModel, AutoTokenizer
import numpy as np
import torch
import ssl
```

```python
model_name = 'deepseek-ai/DeepSeek-OCR'
```

```python

# Tenta ajustar o contexto SSL para ignorar a verificação do certificado (solução temporária)
ssl._create_default_https_context = ssl._create_unverified_context

tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True)
```

```python
# Tenta ajustar o contexto SSL para ignorar a verificação do certificado (solução temporária)
ssl._create_default_https_context = ssl._create_unverified_context

device = "cpu"
dtype = torch.float32

model = AutoModel.from_pretrained(model_name, 
                                  trust_remote_code=True).eval().to(device=device, dtype=dtype)
```

```python
prompt = "<image>\n<|grounding|>Convert the document to markdown"
```

```python
res = model.infer(
    tokenizer,
    prompt = prompt,
    image_file = '',
    base_size=1024, # Native Resolutions: 512,640,1024,1280
    crop_mode=True
)
```
